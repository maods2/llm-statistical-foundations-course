## 1. What is a Jupyter notebook?

## 7. Reading outputs

Outputs appear right below a cell and come in three common shapes:

- **Text / numbers** — printed results, like the messages above.
- **Tables** — neat grids of data (produced by a library called *pandas*).
- **Plots** — charts and figures (produced by *matplotlib*).

Here is a tiny example of a plot so you know what to expect.


````python
import matplotlib.pyplot as plt

x = [0, 1, 2, 3, 4, 5]
y = [0, 1, 4, 9, 16, 25]   # x squared

plt.figure(figsize=(5, 3))
plt.plot(x, y, marker="o")
plt.title("A first plot: y = x squared")
plt.xlabel("x"); plt.ylabel("y")
plt.show()
````

## 8. When something goes wrong (it happens to everyone)

If a cell shows a red **error message**, don't worry — errors are a normal part of computing. Try these, in order:

1. **Re-run the cells above it, from the top.** The most common cause is running cells out of order, so a later cell refers to something an earlier cell hadn't defined yet. In Colab: **Runtime → Run all**.
2. **Restart and run again.** **Runtime → Restart runtime**, then **Runtime → Run all**. This clears everything and starts fresh — a clean slate that fixes most problems.
3. **Read the last line of the error.** It often names the issue in plain-ish English (e.g. *"NameError: name 'x' is not defined"* means a cell above wasn't run).
4. Still stuck? Ask an instructor or a neighbour. You truly cannot damage your computer or the course files.

> 🔑 **The golden rule:** run cells **top to bottom, in order**. About 90% of beginner problems come from skipping a cell.
